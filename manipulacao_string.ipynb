```python
minhaString = "Hello"
 # Acessar caracteres individuais
print(minhaString[0])  # H
print(minhaString[1])  # e
print(minhaString[2])  # l
print(minhaString[3])  # l
print(minhaString[4])  # o

# Concatenar strings
outraString = " World"
novaString = minhaString + outraString
print(novaString)  # Hello World

# Repetir strings
repetida = minhaString * 3
print(repetida)  # HelloHelloHello

# Fatiar strings
print(minhaString[1:4])  # ell
print(minhaString[:3])   # Hel
print(minhaString[2:5])  # llo
print(minhaString[::2])  # Hlo
print(minhaString[-5]) # H
print(minhaString[-1]) # o
print(minhaString[:-1]) # Hell
print(minhaString[::-1]) # olleH
print(minhaString[::1]) # Hello
print(type(minhaString)) # <class 'str'>

print(dir(minhaString)) # metodos disponíveis para strings

print(len(minhaString)) # 5 (número de caracteres na string)
print(len("Hello World")) # 11 (número de caracteres na string)
print(minhaString.upper())  # HELLO
print(minhaString.lower())  # hello
print(minhaString.capitalize())  # Hello
print(minhaString.title())  # Hello
print(minhaString.strip())  # Hello (remove espaços em branco)
print(minhaString.replace("l", "x"))  # Hexxo
print(minhaString.split("e"))  # ['H', 'llo']
print(minhaString.find("o"))  # 4 (índice da primeira ocorrência de "o")
print(minhaString.count("l"))  # 2 (número de ocorrências de "l")

# Verificar se a string começa ou termina com um determinado substring
print(minhaString.startswith("He"))  # True
print(minhaString.startswith("he")) # False (case-sensitive)
print(minhaString.endswith("lo"))  # True
# Verificar se a string contém um determinado substring
print("ell" in minhaString)  # True
print("xyz" in minhaString)  # False
```
